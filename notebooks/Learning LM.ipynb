# Preparing

## Training Data

```python
# Data Loading Code Hidden Here
import pandas as pd

# Load training data from CSV file
melbourne_file_path = '../input/train.csv'
melbourne_data = pd.read_csv(melbourne_file_path)

# Filter rows with missing price values
#filtered_melbourne_data = melbourne_data.dropna(axis=1)

# Choose target and features

melbourne_features = melbourne_data.select_dtypes(exclude=['object']).columns
#melbourne_features = melbourne_features.drop('SalePrice')

#melbourne_features = ['GrLivArea', '1stFlrSF', '2ndFlrSF', 'TotalBsmtSF', 'BsmtFinSF1', 'GarageArea', 'LotArea'] 

#melbourne_features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']
#maybes: 'Neighborhood', 'dwelling', 'OverallCond', 'YearBuilt', 'YearRemodAdd',


X = melbourne_data[melbourne_features].drop('SalePrice', axis=1)
y = melbourne_data.SalePrice
```

## DecisionTreeRegressor model

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)

# Define model
melbourne_DTR_model = DecisionTreeRegressor(max_leaf_nodes=100, random_state=1)
# Fit model
melbourne_DTR_model.fit(train_X, train_y)

# Validation
predicted_home_prices = melbourne_DTR_model.predict(val_X)
mean_absolute_error(val_y, predicted_home_prices)
```

## RandomForestRegressor model

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=1)
melbourne_RFR_model = RandomForestRegressor(max_leaf_nodes=750, random_state=1)
melbourne_RFR_model.fit(train_X, train_y)

# Validation
predicted_home_prices = melbourne_RFR_model.predict(val_X)
mean_absolute_error(val_y, predicted_home_prices)
```

# Leaf Nodes

## Find best max leaf nodes number

```python
from sklearn.ensemble import RandomForestRegressor  # opc1
from sklearn.tree import DecisionTreeRegressor      # opc2
from sklearn.metrics import mean_absolute_error

def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0) # mudar para o modelo desejado
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)
    return mae
```

```python
candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500, 750, 1000]
# Write loop to find the ideal tree size from candidate_max_leaf_nodes
max_mae = {
    "max_leaf":0,
    "mae_value":0,
}

for max_leaf_nodes in candidate_max_leaf_nodes:
    my_mae = get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y)
    if max_leaf_nodes == 5:
        max_mae['mae_value']=my_mae
        max_mae['max_leaf']=max_leaf_nodes 
    if my_mae<max_mae['mae_value']:
        max_mae['mae_value']=my_mae
        max_mae['max_leaf']=max_leaf_nodes

# Store the best value of max_leaf_nodes (it will be either 5, 25, 50, 100, 250 or 500)
best_tree_size = max_mae['max_leaf']
print(best_tree_size)
```
